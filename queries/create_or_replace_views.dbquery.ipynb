```sql
CREATE OR REPLACE VIEW analytics_lakehouse.gold.view_daily_denver_vehicle_listings AS
SELECT
  listing_id AS `Listing ID`,
  dealer_id AS `Dealer ID`,
  vehicle_id AS `Vehicle ID`,
  retrieval_date AS `Retrieval Date`,
  first_seen AS `First Seen`,
  vin AS VIN,
  owner_history AS `Owner History`,
  accident_history AS `Accident History`,
  total_top_options AS `Total Top Options`,
  mileage AS Mileage,
  mpg_city AS `MPG City`,
  mpg_highway AS `MPG Highway`,
  mpg_combined AS `MPG Combined`,
  list_price AS `List Price`,
  original_list_price AS `Original List Price`,
  current_price AS `Current Price`,
  one_price AS `One Price`,
  value_variance AS `Value Variance`,
  value_variance_groups AS `Value Variance Groups`,
  total_price_reduction AS `Total Price Reduction`,
  total_price_history_entries AS `Total Price History Entries`,
  days_on_market AS `Days on Market`,
  total_recorded_services AS `Total Recorded Services`
FROM
  analytics_lakehouse.gold.daily_listings
```
