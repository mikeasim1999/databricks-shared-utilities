```python
def ensure_dq_log_table(catalog, schema):
    dq_log_table = f"{catalog}.{schema}.etl_data_quality_log"
    try:
        spark.sql(f"""
            CREATE TABLE IF NOT EXISTS {dq_log_table} (
                run_id STRING,
                table_name STRING,
                rule_name STRING,
                check_type STRING,
                sql_expression STRING,
                severity STRING,
                result STRING,
                fail_count BIGINT,
                message STRING,
                log_datetime TIMESTAMP
            ) USING DELTA
        """)
        logger.info(f"Ensured DQ log table exists: {dq_log_table}")
    except Exception as e:
        logger.error(f"Failed to create DQ log table {dq_log_table}: {e}")
        raise
```
