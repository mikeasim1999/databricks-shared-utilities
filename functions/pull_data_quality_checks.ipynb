```python
def pull_data_quality_checks(source):
    rules = spark.sql(f"""
        SELECT 
            table_name,
            rule_name, 
            sql_expression, 
            severity,
            check_type
        FROM analytics_lakehouse.{TARGET_SCHEMA}.data_quality_checks
        WHERE data_source = '{source}' AND is_active = TRUE
    """).collect()
 
    if len(rules) == 0:
        print(f"No data quality rules found for {source}")
        return []
    
    return rules
```
