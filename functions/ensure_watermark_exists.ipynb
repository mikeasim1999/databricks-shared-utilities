```python
def ensure_watermark_table(catalog, schema_name):
    """
    Creates Watermark table if it doesn't exist with comprehensive metrics.
    
    Args:
        catalog: Catalog name
        schema_name: Schema name for the log table
    """
    full_table_name = f"{catalog}.{schema_name}.etl_watermark"
    
    try:
        table_exists = spark.catalog.tableExists(full_table_name)
        
        if not table_exists:
            logger.info(f"Creating watermark table: {full_table_name}")
            spark.sql(f"""
                CREATE TABLE IF NOT EXISTS {full_table_name} (
                    table_name STRING,
                    last_modified TIMESTAMP,
                    updated_at TIMESTAMP
                ) USING DELTA
            """)
            logger.info(f"Successfully created {full_table_name}")
        else:
            logger.info(f"Watermark table already exists: {full_table_name}")
            
    except Exception as e:
        logger.error(f"Failed to ensure watermark table: {e}")
        raise
```
