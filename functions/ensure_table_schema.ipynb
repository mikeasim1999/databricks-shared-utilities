```python
def ensure_table_schema(target_table, df):
    """
    Ensures target table exists with correct schema.
    Creates table if it doesn't exist.
    
    Args:
        target_table: Full table name
        df: DataFrame with desired schema
    """
    try:
        if not spark.catalog.tableExists(target_table):
            logger.info(f"Creating new table: {target_table}")
            df.limit(0).write.format("delta").mode("overwrite").saveAsTable(target_table)
            logger.info(f"Successfully created {target_table}")
        else:
            logger.debug(f"Table already exists: {target_table}")
    except Exception as e:
        logger.error(f"Failed to ensure schema for {target_table}: {e}")
        raise
```
