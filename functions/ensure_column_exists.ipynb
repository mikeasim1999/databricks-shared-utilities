```python
def column_exists_in_table(table_name, column_name):
    """
    Check if a column exists in a Delta table.
    
    Args:
        table_name: Full table name (catalog.schema.table)
        column_name: Column name to check
        
    Returns:
        Boolean indicating if column exists
    """
    try:
        columns = spark.table(table_name).columns
        return column_name in columns
    except Exception as e:
        logger.warning(f"Could not check column existence in {table_name}: {e}")
        return False
```
